# Cleaning and processing

```python
df_cleaned = df.copy()
```

```python
df_cleaned.shape
```

```python
df_cleaned.drop_duplicates(inplace=True)
```

```python
df_cleaned.shape
```

```python
df_cleaned.head()
```

```python
df_cleaned['Sex'].value_counts()
```

```python
df_cleaned['Sex'] = df_cleaned['Sex'].map({'M':0,'F':1})
```

```python
df_cleaned.head()
```

```python
df_cleaned['ExerciseAngina'].value_counts()
```

```python
df_cleaned['ExerciseAngina'] = df_cleaned['ExerciseAngina'].map({'N':0,'Y':1})
```

```python
df_cleaned.head()
```

```python
df_cleaned['RestingECG'].value_counts()
```

```python
df_cleaned['ST_Slope'].value_counts()
```

```python
df_cleaned = pd.get_dummies(df_cleaned,columns=['ST_Slope','RestingECG'],drop_first = True,dtype=int)
```

```python
df_cleaned.head()
```

```python
df['ChestPainType'].value_counts()
```

```python
df_cleaned = pd.get_dummies(df_cleaned,columns=['ChestPainType'],drop_first=True,dtype=int)
df_cleaned.head()
```

```python
df_cleaned.rename(columns={
    'Sex':'isFemale',
    'ExerciseAngina':'isExerciseAngina'
},inplace=True)
```

```python
df_cleaned.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, f1_score, classification_report
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
```

```python
models = {
    "Logistic Regression": LogisticRegression(),
    "KNN": KNeighborsClassifier(),
    "Naive Bayes": GaussianNB(),
    "Decision tree": DecisionTreeClassifier(random_state = 42),
    "SVM": SVC()
}
```

```python
X = df_cleaned.drop( columns=['HeartDisease'])
y = df_cleaned['HeartDisease']
X_train,X_test,y_train,y_test = train_test_split(X, y, test_size=0.20, random_state=42)
scaler = StandardScaler()
X_train[numeric_cols] = scaler.fit_transform(X_train[numeric_cols])
X_test[numeric_cols] = scaler.transform(X_test[numeric_cols])
result = []
for name,model in models.items():
    model.fit(X_train,y_train)
    y_pred = model.predict(X_test)
    acc = accuracy_score(y_test,y_pred)
    f1 = f1_score(y_test,y_pred)
    result.append({
        'model': name,
        'Accuracy': round(acc,4),
        'f1 score': round(f1,4)
})
    
```
